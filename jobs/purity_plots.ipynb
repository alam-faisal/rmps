```python
import sys
import os
sys.path.append(os.path.join(os.getcwd(), '..'))
from rmps_purity import *
import pickle
import matplotlib.pyplot as plt
import scienceplots 
```

```python
def set_style():
    plt.style.use(['science', 'grid', 'no-latex'])
    plt.rcParams['image.cmap'] = 'cividis'
    
    plt.rcParams['axes.labelsize'] = 8 
    plt.rcParams['xtick.labelsize'] = 8  
    plt.rcParams['ytick.labelsize'] = 8 
    plt.rcParams['legend.fontsize'] = 8 
    plt.rcParams['lines.linewidth'] = 1.5
    plt.rcParams['savefig.dpi'] = 300
    plt.rcParams['figure.dpi'] = 300 
    
    plt.rcParams['axes.titlesize'] = 10
    
    plt.rcParams['axes.grid'] = True  
    plt.rcParams['axes.spines.top'] = True 
    plt.rcParams['axes.spines.right'] = True
```

```python
set_style()
fig, ax = plt.subplots(figsize=(3.2,2))
colors = plt.cm.plasma(np.linspace(0.2, 0.8, 3))

with open("p_20_purity.pickle", "rb") as f: 
    bond_dims, avgs, stds, hp = pickle.load(f)
    ax.errorbar(bond_dims, avgs, yerr=stds, label="periodic", color=colors[0])
    ax.axhline(hp, linestyle="--", label="Haar", color="black")

with open("rmps_open_even_20.pickle", "rb") as f: 
    bond_dims, avgs, stds = pickle.load(f)
    ax.errorbar(bond_dims, avgs, yerr=stds, label="open, even", color=colors[1])

with open("rmps_open_staggered_20.pickle", "rb") as f: 
    bond_dims, avgs, stds = pickle.load(f)
    ax.errorbar(bond_dims, avgs, yerr=stds, label="open, staggered", color=colors[2])

plt.legend()
plt.ylabel('purity')
plt.xlabel('bond dimension') 
plt.yscale('log') 
#plt.xscale('log') 

plt.title("n = 20, k = 2")
plt.savefig("staggered1.pdf", bbox_inches='tight')
```
