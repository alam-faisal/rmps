```python
import os

def constant_chi_jobs(chi, mps_type, num_samples=40000, n_max=30, distrib='isometric', param=1.0):
    if distrib == 'isometric' and param == 1.0: 
        title = f"run_{mps_type}_{chi}"
    elif distrib != 'isometric' and param == 1.0: 
        title = f"run_{mps_type}_{distrib}_{chi}"
    elif distrib != 'isometric' and param != 1.0: 
        title = f"run_{mps_type}_{distrib}{param}_{chi}"
    
    filename = f"{title}.sbatch"
    
    content = f"""#!/bin/bash -l
#SBATCH --job-name={title}
#SBATCH --nodes=1
#SBATCH --exclusive
#SBATCH --time=2-0:00:00
#SBATCH --no-requeue
#SBATCH --output={title}.out
#SBATCH --error={title}.err
#SBATCH --qos=long
conda activate qaravan
python rmps_purity.py {n_max} 2 {num_samples} {mps_type} {chi} {distrib} {param}
"""
    
    with open(filename, 'w') as f:
        f.write(content)
    print(f"Generated {filename}")
```

```python
mps_type = "ti"
chi_list = [4,6,8,10,12,14,16,20,24,28,32]
for chi in chi_list: 
    constant_chi_jobs(chi, mps_type)
```

```python
mps_type = "open"
chi_list = [4,6,8,10,12,14,16]
distrib = 'uniform'
for chi in chi_list: 
    constant_chi_jobs(chi, mps_type, distrib=distrib)
```

```python
mps_type = "open"
chi_list = [4,6,8,10,12,14,16]
distrib = 'gaussian'
param = 1 
for chi in chi_list: 
    constant_chi_jobs(chi, mps_type, distrib=distrib, param=param)
```
